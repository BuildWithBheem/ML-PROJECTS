```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('/content/arxiv_data.csv')
```

```python
df.nunique()
```

```python
df = df.drop_duplicates()
```

```python
df = df.drop(columns = ['terms'])
```

```python
import re
import nltk
nltk.download('stopwords')
from nltk.corpus import stopwords
from nltk.stem.porter import PorterStemmer

corpus = []

for i in range (0, 10000):
  paper = re.sub('[^a-zA-Z]',' ',df.iloc[i,1])
  paper = paper.lower()
  paper = paper.split()
  ps = PorterStemmer()
  paper = [ps.stem(words) for words in paper]

  paper = ' '.join(paper)

  corpus.append(paper)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfdf = TfidfVectorizer()
x = tfdf.fit_transform(corpus)
y = df.iloc[0:10000,-1]
```

```python
from  sklearn.metrics.pairwise import cosine_similarity
query = "Depth-Estimation"
query = re.sub('[^a-zA-Z]',' ',query)
query = query.lower()
query = query.split()

ps = PorterStemmer()
query = [ps.stem(words) for words in query]
query = ' '.join(query)
query = tfdf.transform([query])
cs = cosine_similarity(query,x).flatten()
tops = cs.argsort()[::-1]
```

```python
print(df.iloc[tops[0]])
```
